# Gikomba Stall Sales and Stock Analysis

Two weeks of sales and stock data analyzed with core Python.

```python
daily_sales = [2300, 1950, 2800, 2100, 3600, 5200, 4100,
               2500, 2200, 3100, 2400, 3900, 5600, 4700]   # 14 days, KES

days = ["Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun"] * 2

# qty = kg in stock, reorder = reorder level (kg), cost/price = KES per kg
stock = {
    "Tomatoes": {"qty": 18, "reorder": 25, "cost": 60, "price": 90},
    "Onions":   {"qty": 40, "reorder": 20, "cost": 70, "price": 100},
    "Sukuma":   {"qty": 9,  "reorder": 15, "cost": 30, "price": 50},
    "Potatoes": {"qty": 55, "reorder": 30, "cost": 50, "price": 75},
    "Avocado":  {"qty": 12, "reorder": 10, "cost": 80, "price": 140},
}
```

## Task 1: 14-Day Sales Summary

I calculate the overall sales statistics and use each value's first list position to identify its day and week.

```python
total_sales = sum(daily_sales)
average_sales = total_sales / len(daily_sales)
highest_sales = max(daily_sales)
lowest_sales = min(daily_sales)
best_index = daily_sales.index(highest_sales)
worst_index = daily_sales.index(lowest_sales)

print(f"Total sales: KES {total_sales:,.2f}")
print(f"Average daily sales: KES {average_sales:,.2f}")
print(f"Highest sales: KES {highest_sales:,.2f} on {days[best_index]}, week {best_index // 7 + 1}")
print(f"Lowest sales: KES {lowest_sales:,.2f} on {days[worst_index]}, week {worst_index // 7 + 1}")
```

## Task 2: Week-on-Week Growth

I sum each seven-day slice, unpack the two totals, and compare week 2 with week 1.

```python
def week_totals(sales):
    """Return total sales for week 1 and week 2."""
    return sum(sales[:7]), sum(sales[7:14])

week_1_total, week_2_total = week_totals(daily_sales)
growth_percentage = (week_2_total - week_1_total) / week_1_total * 100
print(f"Week 1 total: KES {week_1_total:,.2f}")
print(f"Week 2 total: KES {week_2_total:,.2f}")
print(f"Week-on-week growth: {growth_percentage:.1f}%")
```

## Task 3: Classify Daily Sales

I classify each sales amount against the full-period average using a 20% threshold, then count each label in a dictionary.

```python
def classify_day(amount, average):
    """Classify sales as Strong, Normal, or Weak relative to average."""
    if amount > average * 1.2:
        return "Strong"
    if amount < average * 0.8:
        return "Weak"
    return "Normal"

day_categories = {day: classify_day(amount, average_sales) for day, amount in zip(days, daily_sales)}
category_counts = {"Strong": 0, "Normal": 0, "Weak": 0}
for category in day_categories.values():
    category_counts[category] += 1

print("Daily classifications:")
for index, category in enumerate(day_categories.values()):
    print(f"{index + 1:>2}. {days[index]}: {category}")
print(f"Category counts: {category_counts}")
```

## Task 4: Weekday and Weekend Comparison

I use the weekday positions (Monday to Friday) and weekend positions (Saturday and Sunday) from each week to calculate and compare their averages.

```python
weekday_sales = daily_sales[0:5] + daily_sales[7:12]
weekend_sales = daily_sales[5:7] + daily_sales[12:14]
weekday_average = sum(weekday_sales) / len(weekday_sales)
weekend_average = sum(weekend_sales) / len(weekend_sales)
weekend_difference_percentage = (weekend_average - weekday_average) / weekday_average * 100

print(f"Weekday average: KES {weekday_average:,.2f}")
print(f"Weekend average: KES {weekend_average:,.2f}")
print(f"Weekend average is {weekend_difference_percentage:.1f}% higher than weekday average.")
```

## Task 5: Reorder List and Gross Margins

I compare each product's current quantity with its reorder level and calculate gross margin as (price minus cost) divided by price.

```python
def products_to_reorder(stock):
    """Return products at or below their reorder levels."""
    return [product for product, details in stock.items() if details["qty"] <= details["reorder"]]

reorder_list = products_to_reorder(stock)
gross_margins = {
    product: (details["price"] - details["cost"]) / details["price"] * 100
    for product, details in stock.items()
}

print(f"Products at or below reorder level: {', '.join(reorder_list)}")
print("Gross margin by product:")
for product, margin in gross_margins.items():
    print(f"{product}: {margin:.1f}%")
```

## Task 6: Stall Performance Report

I bring the key sales, growth, day-type, classification, and stock results together in a readable report with KES amounts formatted using thousands separators.

```python
print("Stall Performance Report")
print("========================")
print(f"14-day sales total: KES {total_sales:,.2f}")
print(f"Average daily sales: KES {average_sales:,.2f}")
print(f"Week 1 sales: KES {week_1_total:,.2f}")
print(f"Week 2 sales: KES {week_2_total:,.2f} ({growth_percentage:+.1f}% week-on-week)")
print(f"Best day: {days[best_index]} in week {best_index // 7 + 1}, KES {highest_sales:,.2f}")
print(f"Lowest day: {days[worst_index]} in week {worst_index // 7 + 1}, KES {lowest_sales:,.2f}")
print(f"Weekday average: KES {weekday_average:,.2f}; weekend average: KES {weekend_average:,.2f} ({weekend_difference_percentage:.1f}% higher)")
print(f"Day categories: {category_counts}")
print(f"Restock to reorder level: {', '.join(reorder_list)}")
print("Gross margins: " + ", ".join(f"{product} {margin:.1f}%" for product, margin in gross_margins.items()))
```
